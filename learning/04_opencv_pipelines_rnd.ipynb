## Debugging Practice

When something breaks, use this order:

1. Read the error.
2. Identify the exact line.
3. Inspect the input type/shape/value.
4. Reduce the problem to the smallest failing case.
5. Fix it.
6. Run again.
7. Only then ask AI for help if needed.

Do not paste a large unknown solution into the project.

## AI-Assisted Coding Rule

**First attempt:** write it yourself.

**Second step:** ask AI for a hint or explanation.

**Third step:** compare your code with the suggestion.

**Final step:** rewrite the part you now understand.

The goal is not zero AI use.
The goal is **zero black-box code**.
