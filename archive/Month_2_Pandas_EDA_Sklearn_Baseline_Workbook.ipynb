## Debugging Practice

When something breaks, use this order:

1. Read the error.
2. Identify the exact line.
3. Inspect the input type/shape/value (`df.dtypes`, `df.shape`, `X_train.shape`...).
4. Reduce the problem to the smallest failing case.
5. Fix it.
6. Run again.
7. Only then ask AI for help if needed.

Do not paste a large unknown solution into the project.
